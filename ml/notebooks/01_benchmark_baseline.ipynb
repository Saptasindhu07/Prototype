# Independent benchmark and baseline

Offline experiments only. No application/API connections or database writes. The demo is fictional; real topic judgments remain pending. Run cells from top to bottom.

```python
from pathlib import Path
import sys

here = Path.cwd()
PROJECT = next(p for p in [here, *here.parents] if (p / "ml/research_ml").is_dir())
if str(PROJECT / "ml") not in sys.path:
    sys.path.insert(0, str(PROJECT / "ml"))

from research_ml.demo import demo_data
from research_ml.benchmark import run_baseline, evaluate_run
from research_ml.data import load_corpus, read_json
print("Independent ML workspace:", PROJECT / "ml")
```

## Fully judged fictional demo
This checks the pipeline, not real search quality. No external packages or running MongoDB are required.

```python
demo_corpus, demo_benchmark = demo_data()
demo_run = run_baseline(demo_corpus, demo_benchmark)
demo_report = evaluate_run(demo_run, demo_corpus, demo_benchmark)
print("Demo statuses:", demo_report["statuses"])
for row in demo_report["queries"]:
    print(row["query_id"], row["metrics"])
```

## Existing real snapshot, dev split only
The CLI snapshot command is the only component that connects directly to MongoDB. This cell reads its local output. Exact-identifier checks are not semantic relevance scores. The held-out test split is left unused.

```python
corpus_path = PROJECT / "work/ml/corpus-v1.json"
benchmark_path = PROJECT / "work/ml/benchmark-v1.json"

if corpus_path.exists() and benchmark_path.exists():
    corpus = load_corpus(corpus_path)
    benchmark = read_json(benchmark_path)
    baseline_run = run_baseline(corpus, benchmark, split="dev", depth=50)
    baseline_report = evaluate_run(baseline_run, corpus, benchmark)
    print("Documents:", len(corpus["documents"]))
    print("Dev query statuses:", baseline_report["statuses"])
    print("Full saved report:", PROJECT / "work/ml/bm25-dev-v1/report.md")
else:
    print("Real snapshot not found. Use the snapshot/prepare CLI commands in ml/README.md.")
```
